```python
data = [(10, 20, 40), (40, 50, 60), (70, 80, 90)]

new_list = []
for t in data:
    temp = list(t)
    temp[-1] = 100
    new_list.append(tuple(temp))

print(new_list)
```

```python
items = (("item1", 10.5), ("item2", 5.7), ("item3", 8.9))

sorted_items = tuple(sorted(items, key=lambda x: x[1]))

print(sorted_items)
```

```python
words = ["hi", "hello", "fantastic", "world"]

longest = max(words, key=len)

print(longest)
```
