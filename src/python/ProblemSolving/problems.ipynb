```python
def reverse_string(s:str)->str:
    ''' this fun is reverse the string
    Args:
        s (str): The string to reverse

    Returns:
        str: The reversed string
    '''

    lst=s.split()
    newl=lst[::-1]
    return newl
'''فمعناها: هات العناصر كلها ولكن امشِ من الآخر للأول'''
```

```python
result = reverse_string("Hello World from Python")
for word in result:
    print(word, end=" ")
```

```python
def playPanjo(name:str)->str:
    ''' this fun is play panjo game
    Args:
        name (str): The name to play with

    Returns:
        str: The result of the game
    '''
    if(name[0].lower()=='r'):
        return name + " plays banjo" 
    else :
        return name + " does not play banjo"
```

```python
print(playPanjo("rorr"))
```
