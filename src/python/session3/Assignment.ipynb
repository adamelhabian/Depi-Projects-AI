```python
def validation(email:str):
     if email.count('@') !=1 or email.count('.') ==0 :
          return "Invalid email"
     else :
          ind=email.find('@')
          print(f"your user name is {email[:ind]}")
          dot=email.rfind('.')
          print(f"your domain is {email[ind+1:dot]}")
          if email.endswith(".com"):
               print("Commercial Domain")
          elif email.endswith(".edu"):
               print("Educational Domain")
          else:
               print("Other Domain")
          
```

```python
def decode(msg:str)->str:
    s=msg.index("@")
    e=msg.rfind("!!!")
    realmsg=msg[s+1:e]
    words=realmsg.split()
    words[0]=words[0][::-1]
    
    
    words [1]= words[1].replace("I", "")
    words [1]= words[1].replace("O", "")
    words [1]= words[1].replace("U", "")
    words [1]= words[1].replace("E", "")
    words [1]= words[1].replace("A", "")
   
     
    return words
```
