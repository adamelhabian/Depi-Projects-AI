```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"my name is {self.name}, age is {self.age}"

    def __repr__(self) -> str:
        return f"my name is {self.name}, age is {self.age}"
```

```python
s1 = student("test", 10)
str(s1)
s1
```
