```python
class student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"my name is {self.name}, age is {self.age}"

    def __repr__(self) -> str:
        return f"my name is {self.name}, age is {self.age}"
```

```python
s1 = student("test", 10)
str(s1)
s1
```

```python
def arr_reshape(arr:list ,rows:int , cols:int ):
    try:
        arr=np.array(arr)
        return arr.reshape(rows,cols)
    except(TypeError, ValueError) as e:
        raise ValueError(f"invalid reshape operation :{e}")
    
```

```python
vec=[1,2,3,4,5,6]
arr_reshape(vec, 2,3)
```
