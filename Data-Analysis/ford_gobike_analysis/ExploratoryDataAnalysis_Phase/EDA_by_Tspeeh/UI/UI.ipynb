```python
import tkinter as tk
from tkinter import ttk


class FordGoBikeDashboard:
    def __init__(self, root):
        self.root = root

        # =========================
        # Window Configuration
        # =========================
        self.root.title("Ford GoBike Analytics Dashboard")
        self.root.geometry("1400x850")
        self.root.minsize(1100, 700)
        self.root.configure(bg="#F5F7FA")

        # =========================
        # Colors
        # =========================
        self.sidebar_color = "#FFFFFF"
        self.bg_color = "#F5F7FA"
        self.card_color = "#FFFFFF"
        self.primary = "#19A7CE"
        self.secondary = "#7B61FF"
        self.text_color = "#263238"
        self.gray = "#78909C"
        self.border = "#E5E7EB"

        # =========================
        # Main Layout
        # =========================
        self.create_sidebar()
        self.create_main_content()

    # =========================================================
    # SIDEBAR
    # =========================================================

    def create_sidebar(self):

        sidebar = tk.Frame(
            self.root,
            bg=self.sidebar_color,
            width=250
        )

        sidebar.pack(
            side="left",
            fill="y"
        )

        sidebar.pack_propagate(False)

        # ---------- Logo / Title ----------

        logo_frame = tk.Frame(
            sidebar,
            bg=self.sidebar_color
        )

        logo_frame.pack(
            padx=20,
            pady=(25, 10),
            fill="x"
        )

        tk.Label(
            logo_frame,
            text="🚲",
            font=("Arial", 28),
            bg=self.sidebar_color
        ).pack(anchor="w")

        tk.Label(
            logo_frame,
            text="Ford GoBike",
            font=("Arial", 18, "bold"),
            fg=self.text_color,
            bg=self.sidebar_color
        ).pack(anchor="w")

        tk.Label(
            logo_frame,
            text="Analytics Dashboard",
            font=("Arial", 9),
            fg=self.gray,
            bg=self.sidebar_color
        ).pack(anchor="w")

        # ---------- Separator ----------

        ttk.Separator(sidebar).pack(
            fill="x",
            padx=20,
            pady=15
        )

        # ---------- Filters Title ----------

        tk.Label(
            sidebar,
            text="FILTERS",
            font=("Arial", 11, "bold"),
            fg=self.primary,
            bg=self.sidebar_color
        ).pack(
            anchor="w",
            padx=20,
            pady=(5, 15)
        )

        # ---------- Date Range ----------

        self.create_filter_label(
            sidebar,
            "📅  Date Range"
        )

        date_frame = tk.Frame(
            sidebar,
            bg=self.sidebar_color
        )

        date_frame.pack(
            padx=20,
            fill="x"
        )

        self.start_date = ttk.Entry(date_frame)
        self.start_date.insert(0, "01/01/2020")
        self.start_date.pack(
            side="left",
            fill="x",
            expand=True
        )

        tk.Label(
            date_frame,
            text=" → ",
            bg=self.sidebar_color,
            fg=self.gray
        ).pack(side="left")

        self.end_date = ttk.Entry(date_frame)
        self.end_date.insert(0, "31/01/2020")
        self.end_date.pack(
            side="left",
            fill="x",
            expand=True
        )

        # ---------- User Type ----------

        self.create_filter_label(
            sidebar,
            "👤  User Type"
        )

        self.user_type = ttk.Combobox(
            sidebar,
            values=[
                "All",
                "Subscriber",
                "Customer"
            ],
            state="readonly"
        )

        self.user_type.set("All")

        self.user_type.pack(
            padx=20,
            fill="x"
        )

        # ---------- Gender ----------

        self.create_filter_label(
            sidebar,
            "⚥  Gender"
        )

        self.gender = ttk.Combobox(
            sidebar,
            values=[
                "All",
                "Male",
                "Female",
                "Other"
            ],
            state="readonly"
        )

        self.gender.set("All")

        self.gender.pack(
            padx=20,
            fill="x"
        )

        # ---------- Age Group ----------

        self.create_filter_label(
            sidebar,
            "🎂  Age Group"
        )

        self.age_group = ttk.Combobox(
            sidebar,
            values=[
                "All",
                "Young",
                "Adult",
                "Senior"
            ],
            state="readonly"
        )

        self.age_group.set("All")

        self.age_group.pack(
            padx=20,
            fill="x"
        )

        # ---------- Apply Button ----------

        tk.Button(
            sidebar,
            text="Apply Filters",
            font=("Arial", 10, "bold"),
            bg=self.primary,
            fg="white",
            activebackground=self.secondary,
            activeforeground="white",
            relief="flat",
            cursor="hand2",
            padx=10,
            pady=8
        ).pack(
            padx=20,
            pady=25,
            fill="x"
        )

    # =========================================================
    # FILTER LABEL
    # =========================================================

    def create_filter_label(self, parent, text):

        tk.Label(
            parent,
            text=text,
            font=("Arial", 9, "bold"),
            fg=self.text_color,
            bg=self.sidebar_color
        ).pack(
            anchor="w",
            padx=20,
            pady=(18, 6)
        )

    # =========================================================
    # MAIN CONTENT
    # =========================================================

    def create_main_content(self):

        main = tk.Frame(
            self.root,
            bg=self.bg_color
        )

        main.pack(
            side="left",
            fill="both",
            expand=True
        )

        # ---------- Header ----------

        header = tk.Frame(
            main,
            bg=self.bg_color
        )

        header.pack(
            fill="x",
            padx=30,
            pady=(25, 10)
        )

        tk.Label(
            header,
            text="Ford GoBike Interactive Dashboard",
            font=("Arial", 22, "bold"),
            fg=self.text_color,
            bg=self.bg_color
        ).pack(anchor="w")

        tk.Label(
            header,
            text="Explore trip patterns, users and station activity",
            font=("Arial", 10),
            fg=self.gray,
            bg=self.bg_color
        ).pack(anchor="w", pady=(3, 0))

        # ---------- Dashboard Area ----------

        dashboard = tk.Frame(
            main,
            bg=self.bg_color
        )

        dashboard.pack(
            fill="both",
            expand=True,
            padx=30,
            pady=10
        )

        # =====================================================
        # OVERVIEW KPIs
        # =====================================================

        self.section_title(
            dashboard,
            "Overview KPIs"
        )

        kpi_frame = tk.Frame(
            dashboard,
            bg=self.bg_color
        )

        kpi_frame.pack(
            fill="x",
            pady=(0, 15)
        )

        self.create_kpi_card(
            kpi_frame,
            "🚲",
            "183,200",
            "Total Trips"
        )

        self.create_kpi_card(
            kpi_frame,
            "⏱",
            "12.4 min",
            "Avg Duration"
        )

        self.create_kpi_card(
            kpi_frame,
            "👥",
            "24,500",
            "Active Users"
        )

        self.create_kpi_card(
            kpi_frame,
            "📍",
            "Market St",
            "Most Popular Station"
        )

        # =====================================================
        # TIME ANALYSIS
        # =====================================================

        self.section_title(
            dashboard,
            "Time Analysis"
        )

        time_frame = tk.Frame(
            dashboard,
            bg=self.bg_color
        )

        time_frame.pack(
            fill="x",
            pady=(0, 15)
        )

        self.create_chart_card(
            time_frame,
            "Trips by Weekday",
            "📈\n\nWeekday Trips Chart\n\nMon   Tue   Wed   Thu   Fri   Sat   Sun"
        )

        self.create_chart_card(
            time_frame,
            "Trips Over Time",
            "📊\n\nMonthly / Daily Trend\n\nJan   Feb   Mar   Apr   May"
        )

        # =====================================================
        # USER ANALYSIS
        # =====================================================

        self.section_title(
            dashboard,
            "User Analysis"
        )

        user_frame = tk.Frame(
            dashboard,
            bg=self.bg_color
        )

        user_frame.pack(
            fill="x",
            pady=(0, 15)
        )

        self.create_chart_card(
            user_frame,
            "Age Group Analysis",
            "📊\n\nYoung     Adult     Senior"
        )

        self.create_chart_card(
            user_frame,
            "Gender Analysis",
            "📊\n\nMale     Female     Other"
        )

        # =====================================================
        # STATION & TRIP ANALYSIS
        # =====================================================

        self.section_title(
            dashboard,
            "Station & Trip Analysis"
        )

        station_card = tk.Frame(
            dashboard,
            bg=self.card_color,
            highlightbackground=self.border,
            highlightthickness=1
        )

        station_card.pack(
            fill="both",
            expand=True
        )

        tk.Label(
            station_card,
            text="Popular Stations & Trip Flows",
            font=("Arial", 12, "bold"),
            fg=self.text_color,
            bg=self.card_color
        ).pack(
            anchor="w",
            padx=20,
            pady=(15, 5)
        )

        tk.Label(
            station_card,
            text="🗺️\n\nTop Start Stations     →     Top End Stations",
            font=("Arial", 11),
            fg=self.gray,
            bg=self.card_color,
            justify="center"
        ).pack(
            expand=True
        )

    # =========================================================
    # SECTION TITLE
    # =========================================================

    def section_title(self, parent, text):

        tk.Label(
            parent,
            text=text,
            font=("Arial", 13, "bold"),
            fg=self.text_color,
            bg=self.bg_color
        ).pack(
            anchor="w",
            pady=(5, 8)
        )

    # =========================================================
    # KPI CARD
    # =========================================================

    def create_kpi_card(
        self,
        parent,
        icon,
        value,
        title
    ):

        card = tk.Frame(
            parent,
            bg=self.card_color,
            highlightbackground=self.border,
            highlightthickness=1,
            width=220,
            height=90
        )

        card.pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )

        card.pack_propagate(False)

        tk.Label(
            card,
            text=icon,
            font=("Arial", 20),
            fg=self.primary,
            bg=self.card_color
        ).pack(
            side="left",
            padx=(15, 10)
        )

        text_frame = tk.Frame(
            card,
            bg=self.card_color
        )

        text_frame.pack(
            side="left",
            expand=True,
            fill="both",
            pady=15
        )

        tk.Label(
            text_frame,
            text=value,
            font=("Arial", 16, "bold"),
            fg=self.text_color,
            bg=self.card_color
        ).pack(anchor="w")

        tk.Label(
            text_frame,
            text=title,
            font=("Arial", 9),
            fg=self.gray,
            bg=self.card_color
        ).pack(anchor="w")

    # =========================================================
    # CHART CARD
    # =========================================================

    def create_chart_card(
        self,
        parent,
        title,
        placeholder
    ):

        card = tk.Frame(
            parent,
            bg=self.card_color,
            highlightbackground=self.border,
            highlightthickness=1,
            height=190
        )

        card.pack(
            side="left",
            fill="both",
            expand=True,
            padx=5
        )

        card.pack_propagate(False)

        tk.Label(
            card,
            text=title,
            font=("Arial", 11, "bold"),
            fg=self.text_color,
            bg=self.card_color
        ).pack(
            anchor="w",
            padx=15,
            pady=(12, 5)
        )

        tk.Label(
            card,
            text=placeholder,
            font=("Arial", 10),
            fg=self.gray,
            bg=self.card_color,
            justify="center"
        ).pack(
            expand=True
        )


# =============================================================
# RUN APPLICATION
# =============================================================

if __name__ == "__main__":

    root = tk.Tk()

    app = FordGoBikeDashboard(root)

    root.mainloop()
```
